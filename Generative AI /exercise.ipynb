```python
import os
import time

from google import genai

client = genai.Client(api_key=os.environ["GEMINI_API_KEY"])

model = "gemini-3.5-flash"

PROMPT = """
Explain why a database index can make a query faster, and give one example
where adding an index could actually make a system worse. Keep the answer
under 120 words.
""".strip()

start = time.perf_counter()

interaction = client.interactions.create(
    model=model,
    input=PROMPT,
)

latency = time.perf_counter() - start

print(f"Model: {model}")
print(f"Latency: {latency:.3f} seconds")
print(f"Usage: {interaction.usage}")

print("\nAnswer:")
print(interaction.output_text)
```
